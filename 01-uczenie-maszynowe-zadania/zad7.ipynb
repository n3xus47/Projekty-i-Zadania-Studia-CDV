```python
import pandas as pd
import requests
from io import StringIO

# Scraping z nagłówkiem User-Agent
url = "https://en.wikipedia.org/wiki/List_of_tallest_buildings"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)

# Wczytanie właściwej tabeli i uproszczenie nagłówków MultiIndex
df = pd.read_html(StringIO(response.text))[1]
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(1)

# Eksport do CSV i H5
df.to_csv("wiezowce.csv", index=False)
df.astype(str).to_hdf("wiezowce.h5", key="tallest", mode="w")

# Import i wyświetlenie
df_import = pd.read_hdf("wiezowce.h5", key="tallest")
df_import
```
