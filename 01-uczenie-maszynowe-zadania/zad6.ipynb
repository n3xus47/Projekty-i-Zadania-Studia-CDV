```python
import pandas as pd

data = {
    "województwo": ["dolnośląskie", "kujawsko-pomorskie", "lubelskie", "lubuskie", "łódzkie", "małopolskie", "mazowieckie"],
    "nr rejestru": [2, 4, 6, 8, 10, 12, 14],
    "miasto wojewódzkie": ["Wrocław", "Bydgoszcz / Toruń", "Lublin", "Gorzów Wielkopolski / Zielona Góra", "Łódź", "Kraków", "Warszawa"],
    "powierzchnia [km²]": ["19 946,70", "17 971,34", "25 122,46", "13 987,93", "18 218,95", "15 182,79", "35 558,47"],
    "ludność": ["2 904 198", "2 086 210", "2 139 726", "1 018 084", "2 493 603", "3 372 618", "5 349 114"]
}

df = pd.DataFrame(data)

df["powierzchnia_n"] = df["powierzchnia [km²]"].str.replace(" ", "").str.replace(",", ".").astype(float)
wynik = df[df["powierzchnia_n"] < 20000]
print("Miasta wojewódzkie o powierzchni mniejszej niż 20000:\n")
print(wynik[["miasto wojewódzkie", "powierzchnia [km²]"]])
print("\n" * 2)


df["ludnosc_n"] = df["ludność"].str.replace(" ", "").astype(int)
wynik2 = df[df["ludnosc_n"] > 2000000].copy()
print("Województwa z ludnością powyżej 2mln:\n")
print(wynik2[["województwo", "ludność"]])
print("\n" * 2)


nowy_wiersz = pd.DataFrame({
    "województwo": "wielkopolskie",
    "nr rejestru": 30,
    "miasto wojewódzkie": "Poznań",
    "powierzchnia [km²]": "29 826,50",
    "ludność": "3 475 323"
}, index=[7])

df = pd.concat([df, nowy_wiersz])


df["powierzchnia_n"] = df["powierzchnia [km²]"].str.replace(" ", "").str.replace(",", ".").astype(float)
df["ludnosc_n"] = df["ludność"].str.replace(" ", "").astype(int)


df = df.sort_values(by="ludnosc_n", ascending=False)


cols = [c for c in df.columns if c != "miasto wojewódzkie"] + ["miasto wojewódzkie"]
df = df[cols]


df["województwo"] = df["województwo"].str.capitalize()


ser_gestosc = pd.Series((df["ludnosc_n"] / df["powierzchnia_n"]) > 140)
ser_gestosc.index = df["województwo"]


df = df[df["województwo"] != "Lubuskie"]


print("Parametry statystyczne (describe):\n")
print(df[["powierzchnia_n", "ludnosc_n"]].describe())
print("\n" * 2)


pd.set_option("display.expand_frame_repr", False)
pd.set_option("display.width", 1200)
print("FINALNY DATAFRAME:\n")
print(df.drop(columns=["powierzchnia_n", "ludnosc_n"]))
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({"a": [1, np.nan, 3], "b": [np.nan, 5, 6]})

#to isna szuka dziur
print(df.isna().sum(axis=1))
print("\n" * 2)

# fillna uzupelnia dziury, w tym przypadku srednia, czyli mean
df = df.fillna(df.mean())
print(df)
```
