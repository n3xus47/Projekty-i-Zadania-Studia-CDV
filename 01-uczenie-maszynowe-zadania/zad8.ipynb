```python
import matplotlib.pyplot as plt

klasy = ["A", "B", "C", "D"]
uczniowie = [25, 18, 30, 22]

plt.bar(klasy, uczniowie, color="skyblue")

plt.title("Liczba uczniów w klasach")
plt.xlabel("Klasa")
plt.ylabel("Liczba uczniów")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = x**2

plt.plot(x, y, label="y = x²", color="red")

plt.title("Wykres funkcji kwadratowej")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

dni = pd.date_range(start="2023-10-01", end="2023-11-30", freq="D")
# Przykładowe wartości indeksu (start ok. 1900 pkt, trend wzrostowy)
notowania = np.linspace(1900, 2250, len(dni)) + np.random.normal(0, 30, len(dni))

plt.figure(figsize=(12, 6))
plt.plot(dni, notowania, label="WIG20", color="green", linewidth=1.5)


plt.plot(dni[14], notowania[14], marker="o", markersize=10, color="red", label="Wydarzenie 1")
plt.plot(dni[45], notowania[45], marker="s", markersize=10, color="blue", label="Wydarzenie 2")

plt.title("Notowania indeksu WIG20 (10.2023 - 11.2023)")
plt.xlabel("Data")
plt.ylabel("Punktacja")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Heart_Disease_Prediction.csv')

# wybor kolumn z liczbami
numeric_df = df.select_dtypes(include=['number'])
corr_matrix = numeric_df.corr()

# rysowanie mapy ciepła z seaborn
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Zadanie 4: Macierz korelacji")
plt.show()
```
