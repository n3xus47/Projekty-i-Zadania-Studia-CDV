```python
import numpy as np
import matplotlib.pyplot as plt



rng = np.random.default_rng()

wartosci = rng.choice([-1,1], size=100)

tablica = np.concatenate([[0], np.cumsum(wartosci)])

print(f"min. wartosc = {np.min(tablica)}")
print(f"max. wartosc = {np.max(tablica)}")

#rysownanie z tego calego matplotlib
plt.figure(figsize = (10,6))
plt.plot(tablica)
plt.xlabel("wartosc na osi x")
plt.ylabel("pozycja na osi y")
plt.axhline(0, color="red")
plt.show()


#50x petla
licznik = 0
plt.figure(figsize = (10,6))

for _ in range(50):
    nowe_wartosci = rng.choice([-1,1], size=100)
    nowa_tablica = np.concatenate([[0], np.cumsum(nowe_wartosci)])

    if np.any(nowa_tablica > 30):
        licznik += 1

    plt.plot(nowa_tablica, alpha = 0.4) # dla widocznosci zeby nie bylo spaghetti

plt.axhline(30, color="red", linewidth=2, label="Próg 30")
plt.title(f"50 symulacji (Przekroczono 30: {licznik} razy)")
plt.xlabel("Krok")
plt.ylabel("Pozycja")
plt.legend()
plt.show()

print(f"liczba symulacji z 50 ktore przekroczyly 30 = {licznik}")
```
