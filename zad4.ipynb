```python
city_list = []

city_list_k = []

city_list_6 = []

city_list.append("Warszawa")
city_list.append("Krakow")
city_list.append("Poznan")
city_list.append("Gdansk")
city_list.append("Lublin")

print(f"miasta: {city_list}")

for city in city_list:
    if city.lower()[0] > "k":
        city_list_k.append(city)
print(f"miasta alfabetycznie po literze k: {city_list_k}")

for city in city_list:
    if len(city) > 6:
        city_list_6.append(city)
print(f"miasta nazwa dluzsze niz 6 znakow: {city_list_6}")
```

```python
months_dict = {
    1: "styczeń",
    2: "luty",
    3: "marzec",
    4: "kwiecień",
    5: "maj",
    6: "czerwiec",
    7: "lipiec",
    8: "sierpień",
    9: "wrzesień",
    10: "październik",
    11: "listopad",
    12: "grudzień"
}

num_month = int(input("podaj nr miesiaca od 1 do 12: "))

if num_month in months_dict:
    print(f"{num_month} to {months_dict[num_month]}")
else:
    print(f"blad, liczba musi byc w przedziale 1-12 i byc liczba calkowita")


```

```python
import math
r = int(input("podaj promien: "))
def kolo(r):
    pole = math.pi * r**2
    obwod = 2 * math.pi * r
    return (pole, obwod)

wynik_pole, wynik_obwod = kolo(r)

print(f"pole kola {wynik_pole}")
print(f"obwod kola {wynik_obwod}")
```

```python
file = open("text.txt", "w")

while True:
    tekst = input("podaj tekst: ")

    if tekst == "":
        break

# n od new line zeby wszystko nie bylo obok siebie
    file.write(tekst + "\n")

file.close()
```

```python
file = open("people.txt", "w")
while True:
    imie = input("podaj imie: ")
    if imie == "":
        break
    nazwisko = input("podaj nazwisko: ")
    rok_urodzenia = input("podaj rok urodzenia: ")

    file.write(f"{imie}, {nazwisko}, {rok_urodzenia}\n")
file.close()

file = open("people.txt", "r")
for x, line in enumerate(file, 1):
    #formatowanie do tego co bylo w poleceniu
    dane = line.strip().split(",")
    if len(dane) == 3:
        imie, nazwisko, rok_urodzenia = dane
        wiek = 2026 - int(rok_urodzenia)
        print(f"{x}. Imię: {imie}, Nazwisko: {nazwisko}, Wiek: {wiek}")
file.close()
```
