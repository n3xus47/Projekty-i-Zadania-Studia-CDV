```python
# Simple is better than complex

#simple
def suma_s(numbers):
    return sum(numbers)

#complex
def suma_c(numbers):
    total = 0
    for number in numbers:
        total += number
    return total
```

```python
def fib(n):
    if n <= 0:
        return 0
    elif n == 1 or n == 2:
        return 1

    return fib(n-1) + fib(n-2)

print (fib(23))
```

```python
class Square:
    def __init__(self, bok):
        self.bok = bok

    def __str__(self):
        pole = self.bok ** 2
        return f"{pole}"

print(Square(3))
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def calculation(self):
        return self.width * self.height

class Square(Rectangle):
    def __init__(self, side):

        super().__init__(side, side)



rectangle = Rectangle(9, 21)
print(f"pole prostokata 9x21 to: {rectangle.calculation()}")

square = Square(17)
print(f"pole kwadratu o boku 17 to: {square.calculation()}")
```
